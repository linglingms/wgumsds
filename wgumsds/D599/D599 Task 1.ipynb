Data Cleaning

```python
import pandas as pd

# Load the Employee Turnover Dataset (CSV file)
df = pd.read_csv("Employee Turnover Dataset.csv")

# Display the first few rows of the dataset
print("First few rows of the dataset:")
print(df.head())

# 1. Check for duplicate entries
duplicates = df[df.duplicated()]
print(f"\nNumber of duplicate rows: {len(duplicates)}")
if len(duplicates) > 0:
    print("Duplicate rows:")
    print(duplicates)

# 2. Check for missing values
missing_values = df.isnull().sum()
print("\nMissing values in each column:")
print(missing_values)

# 3. Check for inconsistent entries
print("\nUnique values in each column:")
for column in df.columns:
    print(f"{column}: {df[column].unique()}")

# 4. Check for formatting errors
print("\nData types of each column:")
print(df.dtypes)

# 5. Check for outliers in numeric columns
print("\nSummary statistics for numeric columns:")
print(df.describe())
```

```python
# Check for outliers in numeric columns
print("\nSummary statistics for numeric columns:")
print(df.describe())

# Count outliers using the IQR method
print("\nOutlier counts for numeric columns:")
numeric_columns = df.select_dtypes(include=['float64', 'int64']).columns  # Select numeric columns
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)  # First quartile (25th percentile)
    Q3 = df[column].quantile(0.75)  # Third quartile (75th percentile)
    IQR = Q3 - Q1  # Interquartile range
    lower_bound = Q1 - 1.5 * IQR  # Lower bound for outliers
    upper_bound = Q3 + 1.5 * IQR  # Upper bound for outliers

    # Identify outliers
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    outlier_count = len(outliers)

    # Print the count of outliers for the column
    print(f"Column '{column}' has {outlier_count} outliers.")
```

```python
# Check for formatting errors
print("\nFormatting errors in the dataset:")

# 1. Check for incorrect data types
print("\nIncorrect data types:")
for column in df.columns:
    if df[column].dtype == 'object':  # Check for string columns
        try:
            pd.to_numeric(df[column])  # Try converting to numeric
        except ValueError:
            print(f"Column '{column}' contains non-numeric values.")

# 2. Check for unexpected characters or symbols
print("\nUnexpected characters or symbols:")
for column in df.columns:
    if df[column].dtype == 'object':  # Check for string columns
        unexpected_chars = df[column].str.contains(r'[^\w\s]', na=False).sum()  # Regex for non-alphanumeric characters
        print(f"Column '{column}' has {unexpected_chars} entries with unexpected characters.")

# 3. Check for leading or trailing spaces
print("\nWhitespace issues:")
for column in df.columns:
    if df[column].dtype == 'object':  # Check for string columns
        whitespace_issues = df[column].str.contains(r'^\s|\s$', na=False).sum()  # Regex for leading/trailing spaces
        print(f"Column '{column}' has {whitespace_issues} entries with leading or trailing spaces.")

# 4. Check for inconsistent capitalization
print("\nCase sensitivity issues:")
for column in df.columns:
    if df[column].dtype == 'object':  # Check for string columns
        unique_values = df[column].dropna().unique()
        lower_case_values = set(value.lower() for value in unique_values)
        if len(unique_values) != len(lower_case_values):
            print(f"Column '{column}' has inconsistent capitalization.")
```

```python
# 3. Check for inconsistent entries
print("\nUnique values in each column:")
for column in df.columns:
    print(f"{column}: {df[column].unique()}")

# Define valid entries for specific columns (if applicable)
valid_entries = {
    'Department': ['HR', 'IT', 'Finance', 'Marketing', 'Operations'],  # Example valid departments
    # Add more columns and their valid entries if needed
}

# Check for inconsistent entries
print("\nInconsistent entries in each column:")
for column in df.columns:
    if column in valid_entries:  # Check only for columns with predefined valid entries
        inconsistent = ~df[column].isin(valid_entries[column])  # Identify invalid rows
        inconsistent_count = inconsistent.sum()  # Count invalid rows
        print(f"{column}: {inconsistent_count} inconsistent entries")
    else:
        print(f"{column}: No predefined valid entries to check for inconsistencies")
```

# Cleaning the Dataset

```python
import pandas as pd

# Load the Employee Turnover Dataset (CSV file)
df = pd.read_csv("Employee Turnover Dataset.csv")

# 1. Remove Duplicate Rows
print(f"Number of duplicate rows before cleaning: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"Number of duplicate rows after cleaning: {df.duplicated().sum()}")

# 2. Handle Missing Values
print("\nMissing values before cleaning:")
print(df.isnull().sum())

# Fill missing values for numeric columns with the median
numeric_columns = df.select_dtypes(include=['float64', 'int64']).columns
for column in numeric_columns:
    df[column].fillna(df[column].median(), inplace=True)

# Fill missing values for categorical columns with the mode
categorical_columns = df.select_dtypes(include=['object']).columns
for column in categorical_columns:
    df[column].fillna(df[column].mode()[0], inplace=True)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

# 3. Fix Inconsistent Entries
# Example: Standardize entries in the 'Department' column
if 'Department' in df.columns:
    valid_departments = ['HR', 'IT', 'Finance', 'Marketing', 'Operations']
    df['Department'] = df['Department'].str.strip().str.title()  # Remove spaces and standardize capitalization
    df.loc[~df['Department'].isin(valid_departments), 'Department'] = 'Other'  # Replace invalid entries with 'Other'

# 4. Resolve Formatting Errors
# Remove leading/trailing spaces in string columns
for column in df.select_dtypes(include=['object']).columns:
    df[column] = df[column].str.strip()

# Convert date columns to datetime format (if applicable)
if 'Hire Date' in df.columns:
    df['Hire Date'] = pd.to_datetime(df['Hire Date'], errors='coerce')

# Convert numeric columns stored as strings to numeric
for column in df.columns:
    if df[column].dtype == 'object':
        try:
            df[column] = pd.to_numeric(df[column], errors='coerce')
        except ValueError:
            pass

# 5. Handle Outliers
# Use the IQR method to cap outliers
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df[column] = df[column].clip(lower=lower_bound, upper=upper_bound)

# Display the cleaned dataset
print("\nCleaned Dataset:")
print(df.head())

# Save the cleaned dataset to a new CSV file
df.to_csv("Cleaned_Employee_Turnover_Dataset.csv", index=False)
print("\nCleaned dataset saved as 'Cleaned_Employee_Turnover_Dataset.csv'")
```
